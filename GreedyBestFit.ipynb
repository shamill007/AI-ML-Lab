```python
# Define the graph using an adjacency list (dictionary)
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

# Define the start and goal nodes
start_node = 'A'
goal_node = 'F'

# Display the graph and search parameters
print("Graph representation:")
for node, neighbors in graph.items():
    print(f"Node {node} is connected to: {neighbors}")

print(f"\nStart Node: {start_node}")
print(f"Goal Node: {goal_node}")
```

```python
# Define the heuristic values h(n) for each node
# This represents the estimated distance to the goal node 'F'
heuristics = {
    'A': 6,
    'B': 4,
    'C': 2,
    'D': 5,
    'E': 1,
    'F': 0
}

# Display the heuristic values
print("Heuristic values (estimated distance to goal 'F'):")
for node, value in heuristics.items():
    print(f"h({node}) = {value}")
```

```python
def greedy_best_first_search(graph, heuristics, start, goal):
    # Priority queue stores tuples of (heuristic_value, node)
    # We start with our start node
    queue = [(heuristics[start], start)]

    # Keep track of visited nodes to avoid repeating them
    visited = set()

    # Order of node visits for our record
    visited_order = []

    # Dictionary to store the parent of each node to reconstruct the path later
    parent = {start: None}

    while queue:
        # Sort queue based on heuristic value (ascending) so the smallest h(n) is at index 0
        queue.sort(key=lambda x: x[0])

        # Pop the node with the lowest heuristic value
        current_heuristic, current_node = queue.pop(0)

        # If we have already visited this node, skip it
        if current_node in visited:
            continue

        # Mark as visited
        visited.add(current_node)
        visited_order.append(current_node)

        # Check if we have reached the goal
        if current_node == goal:
            break

        # Explore neighbors
        for neighbor in graph[current_node]:
            if neighbor not in visited:
                # Record the parent of the neighbor
                parent[neighbor] = current_node
                # Add to queue with its heuristic value
                queue.append((heuristics[neighbor], neighbor))

    # Reconstruct path from goal to start
    path = []
    current = goal
    while current is not None:
        path.append(current)
        current = parent.get(current)
    path.reverse()  # Reverse to make it start -> goal

    return visited_order, path

# Run the search algorithm
visited_order, final_path = greedy_best_first_search(graph, heuristics, start_node, goal_node)
```

```python
# Print the output of the experiment
print("=== EXPERIMENT RESULTS ===")
print(f"Order of nodes visited: {' -> '.join(visited_order)}")
print(f"Final path found:       {' -> '.join(final_path)}")
```
